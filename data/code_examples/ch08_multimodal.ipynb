# 08 多模态大模型 · 代码示例

**章节**: 08_multimodal_llm.md
**学习目标**: 模拟「把同一知识点转成多种资源描述」的流程
**A3 关联**: A3 系统的多模态资源生成(文字/思维导图/语音/代码)

````python
from collections import Counter
import re

KNOWLEDGE_POINT = {
    "name": "Transformer 自注意力",
    "definition": "Self-Attention 通过 Q/K/V 三个向量计算序列中任意两个位置的关联度。",
    "analogy": "想象一个鸡尾酒会,你能自动过滤噪声,聚焦当前说话的人。",
    "formula": "Attention(Q,K,V) = softmax(QKᵀ/√d_k)V"
}

def gen_text_explanation(kp):
    return f"{kp['name']}:{kp['definition']}"

def gen_mindmap_code(kp):
    return f"""```mermaid
mindmap
  root(({kp['name']}))
    定义
      ::icon(fa fa-book)
      {kp['definition'][:20]}
    直觉
      {kp['analogy'][:20]}
    公式
      {kp['formula']}
```"""

def gen_quiz(kp):
    return {
        "stem": f"请解释 {kp['name']} 的核心思想。",
        "bloom": "Understand",
        "answer": kp['definition']
    }

def gen_audio_script(kp):
    return f"【停顿】同学们好,今天我们学习 {kp['name']}。{kp['analogy']}【停顿】"
````

### 解读

- `gen_text_explanation`: 生成文字讲解
- `gen_mindmap_code`: 生成 Mermaid 思维导图代码(可在前端渲染)
- `gen_quiz`: 生成 Bloom 层级题目
- `gen_audio_script`: 生成 TTS 朗读脚本(对应 voice_scripts/)

````python
print("=== 文字讲解 ===")
print(gen_text_explanation(KNOWLEDGE_POINT))

print("\n=== 思维导图代码 ===")
print(gen_mindmap_code(KNOWLEDGE_POINT))

print("\n=== 题目 ===")
print(gen_quiz(KNOWLEDGE_POINT))

print("\n=== 语音脚本 ===")
print(gen_audio_script(KNOWLEDGE_POINT))
````
