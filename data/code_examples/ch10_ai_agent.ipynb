**学习目标**: 模拟一个 Plan-and-Execute 风格的简单 Agent
**A3 关联**: 模拟 A3 系统 Planner Agent 的核心调度逻辑

```python
class SimpleAgent:
    """Plan-and-Execute 风格的简单 Agent"""
    def __init__(self, name):
        self.name = name
        self.memory = []

    def plan(self, goal):
        """把目标拆成步骤"""
        if "讲解" in goal:
            return ["检索知识库", "生成讲解", "自评质量"]
        elif "出题" in goal:
            return ["读取掌握度", "选 Bloom 层级", "生成题目"]
        else:
            return ["理解目标", "执行"]

    def execute(self, step):
        """执行一步,返回结果"""
        result = f"[{self.name}] 已完成: {step}"
        self.memory.append(result)
        return result

    def run(self, goal):
        print(f"\n[{self.name}] 收到目标: {goal}")
        plan = self.plan(goal)
        print(f"[{self.name}] 计划: {plan}")
        for step in plan:
            result = self.execute(step)
            print(f"  -> {result}")
        return self.memory
```

### 解读

- `plan(goal)`: 根据目标字符串匹配不同的规划模板
- `execute(step)`: 模拟执行一步,把结果记入 memory
- `run(goal)`: 端到端跑完一个目标

```python
tutor = SimpleAgent("Tutor")
result = tutor.run("请讲解 Transformer 注意力机制")
print("\n最终 memory:", result)
```
