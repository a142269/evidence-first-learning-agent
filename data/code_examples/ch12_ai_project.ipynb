**学习目标**: 跑通 A3 系统的最小闭环伪代码
**A3 关联**: 这是 A3 系统端到端流水线的代码骨架

```python
class A3MinimumLoop:
    """A3 个性化学习系统的最小可演示闭环"""
    def __init__(self, student_id):
        self.student_id = student_id
        self.profile = {"transformer": 0.5, "rag": 0.3}

    def retrieve(self, query):
        return f"[知识库召回] {query} 相关的章节内容"

    def generate_explanation(self, query, context, profile):
        return f"[Tutor 讲解] 基于 {context},学生当前掌握度 {profile}"

    def generate_quiz(self, topic, level):
        return {"stem": f"请简述 {topic} 的核心思想", "level": level}

    def evaluate(self, quiz, answer):
        is_correct = "简述" in answer or len(answer) > 5
        return is_correct, ("回答正确" if is_correct else "未答到要点")

    def update_profile(self, topic, delta):
        self.profile[topic] = max(0, min(1, self.profile.get(topic, 0.5) + delta))

    def process_query(self, query):
        # 1. 画像读取
        # 2. 知识库检索
        context = self.retrieve(query)
        # 3. 个性化讲解
        explanation = self.generate_explanation(query, context, self.profile)
        # 4. 推送练习
        quiz = self.generate_quiz("transformer", self.profile["transformer"])
        return {"explanation": explanation, "quiz": quiz}
```

### 解读

A3 的最小闭环包含 6 步:画像 → 检索 → 讲解 → 出题 → 评估 → 画像更新

本代码骨架省略了 LLM 调用(用字符串模拟),但保留了完整的数据流。
在真实 A3 系统中,retrieve/evaluate/explanation 内部都会调用大模型 + RAG。

```python
a3 = A3MinimumLoop(student_id="demo-student-1")

# 模拟一次完整交互
result = a3.process_query("我不懂注意力机制")
print("【讲解】:", result["explanation"])
print("【题目】:", result["quiz"])

# 模拟学生答题
is_correct, feedback = a3.evaluate(result["quiz"], "注意力机制就是 QKV 三向量")
print("\n【评估】:", "正确" if is_correct else "错误", "-", feedback)

# 画像更新
delta = +0.1 if is_correct else -0.05
a3.update_profile("transformer", delta)
print("\n【新掌握度】 transformer =", a3.profile["transformer"])
```
